**Sistema de Gestão de Notas**

```python
# --- FUNÇÕES ---

def calcular_media(notas):
    """Soma as notas da lista e divide pela quantidade."""
    return round(sum(notas) / len(notas), 2)

def verificar_situacao(media):
    """Define a situação com base na média 7."""
    return "Aprovado" if media >= 7 else "Reprovado"
```

```python
# --- EXECUÇÃO DO PROGRAMA ---

print("=== GESTÃO DE NOTAS ===")

# Loop para garantir uma quantidade válida de notas
while True:
    try:
        qtd_notas_ano = int(input("Quantas notas haverá no ano? "))

        if qtd_notas_ano <= 0:
            print("[ERRO] A quantidade de notas deve ser maior que zero! Tente novamente.\n")
        else:
            break  # Sai do loop assim que uma quantidade válida for inserida

    except ValueError:
        print("[ERRO] Digite apenas números inteiros válidos! Tente novamente.\n")

lista_de_notas = []  # lista para armazenar as notas do aluno

# faz o loop para registrar as notas do aluno na lista
print("\n--- Cadastro de Notas ---")
for i in range(qtd_notas_ano):
    while True:
        try:
            nota = float(input(f"Digite a {i + 1}ª nota: "))
            if 0 <= nota <= 10:
                lista_de_notas.append(nota)
                break  # Nota válida inserida, passa para a próxima
            else:
                print("[ERRO] A nota deve estar entre 0 e 10! Tente novamente.")
        except ValueError:
            print("[ERRO] Digite um número válido para a nota (ex: 8 ou 7.5)!\n")

# Processamento através das funções
media_final = calcular_media(lista_de_notas)
situacao = verificar_situacao(media_final)

# RELATÓRIO FINAL
print("\n" + "="*30)
print("      RELATÓRIO FINAL      ")
print("="*30)
print(f"Notas digitadas: {lista_de_notas}")
print(f"Média do aluno:  {media_final}")
print(f"Situação:        {situacao}")
print("="*30)
```
